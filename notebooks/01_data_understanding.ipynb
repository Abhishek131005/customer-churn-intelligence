## Objective

This notebook performs the initial assessment of the IBM Telco Customer Churn dataset.

The goal is to understand the structure, features, target variable, data types, missing values, duplicates, categorical values, and potential data-quality issues before performing any cleaning or feature engineering.

No modelling decisions are made at this stage.

```python
from pathlib import Path

import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None)
```

```python
PROJECT_ROOT = Path.cwd().parent

DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "WA_Fn-UseC_-Telco-Customer-Churn.csv"
)

df = pd.read_csv(DATA_PATH)

print(f"Dataset shape: {df.shape}")
df.head()
```

```python
schema = pd.DataFrame({
    "column": df.columns,
    "dtype": df.dtypes.astype(str).values,
    "non_null": df.notna().sum().values,
    "null_count": df.isna().sum().values,
    "unique_values": df.nunique().values
})

schema
```

```python
for column in df.columns:
    print(f"\n{'=' * 60}")
    print(f"Column: {column}")
    print(f"Data type: {df[column].dtype}")
    print(f"Unique values: {df[column].nunique()}")
    
    if df[column].nunique() <= 20:
        print(df[column].value_counts(dropna=False))
```

```python
missing = (
    df.isnull()
    .sum()
    .sort_values(ascending=False)
    .to_frame("missing_count")
)

missing["missing_percentage"] = (
    missing["missing_count"] / len(df) * 100
)

missing
```

```python
blank_counts = {}

for column in df.select_dtypes(include="object").columns:
    blank_counts[column] = (
        df[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

pd.Series(blank_counts).sort_values(ascending=False)
```

## Initial Data Quality Findings

The initial inspection identified several points that require attention during the data-cleaning phase:

1. `customerID` is a unique identifier and should not be used directly as a predictive feature.
2. `TotalCharges` requires further inspection because it is represented as an object/string column despite conceptually being numerical.
3. Missing-value analysis must include blank or whitespace values, not only standard null values.
4. `SeniorCitizen` is stored numerically but represents a categorical binary attribute.
5. The target variable `Churn` is not perfectly balanced, so model evaluation should not rely solely on accuracy.
6. Service-related categorical features contain values such as `No internet service` and `No phone service`, which carry semantic information and should not automatically be treated as ordinary missing values.

No transformations have been applied yet. Cleaning decisions will be made in the next stage after investigating these issues.
